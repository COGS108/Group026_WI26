## Research Question

How does public transportation usage (buses and metro), smoking prevalence rates, and electric vehicle (EV) adoption influence air quality and the rate of reported asthma attacks among asthma patients in Houston, San Francisco, and New York City?

## Background and Prior Work

### Background:
Air pollution is one of the leading environmental health risks worldwide, and urban transportation is a major contributor. Private vehicles, such as cars, trucks, and motorcycles, emit pollutants such as particulate matter (PM2.5 and PM10), nitrogen dioxide (NO2), carbon monoxide (CO), and volatile organic compounds (VOCs), all of which have been strongly linked to respiratory diseases, including asthma and chronic obstructive pulmonary disease (COPD). (1, 2)

From the report of Asthma and Allergy Foundation of America (5), poor air quality can negatively affect everyone’s health. Research shows that air pollution can make asthma worse and trigger asthma symptoms. Air pollution includes gases, smoke from fires, volcanic ash, dust particles, emissions from transportation, and other substances that can harm the lungs.

As cities grow, the concentration of vehicles increases, intensifying both traffic congestion and emissions, which has direct implications for public health. Understanding the link between transportation and air quality is crucial for policy-making aimed at improving population health.

Public transportation, such as buses, metros, and high-speed rail (HSR), has long been proposed as a solution to reduce private vehicle use. Expanding public transit can enable ease of movement for lower-income individuals while lowering urban air pollution by reducing congestion and moving more people collectively without increasing operational emissions per person. (3)

Other factors besides transportation also influence urban air quality and lung health, including smoking rates and adoption of EVs, which can have both positive and negative effects.

Our project aims to specifically observe the link between public transportation systems and air quality and lung disease, controlling for the factors mentioned above. Some of the cities we will investigate include Houston, New York City, and San Francisco.

It is important to note that our study is observational and relies on publicly available datasets, so we cannot definitively prove causation between public transportation and air quality. Many of the factors mentioned above are interrelated, and unobservable variables may influence the results. To mitigate bias, we will examine multiple cities and control for known confounding factors as carefully as possible, aiming to produce robust, evidence-based insights.

### Air Quality Data

This dataset looks at air pollution trends in U.S. cities (called Core Based Statistical Areas, or CBSAs) from 2000 to 2023. Each entry shows a pollutant measured in a specific city along with a statistic that summarizes how that pollutant is evaluated. For instance, one major metric is ozone (O3), which is reported using the “4th Max” value — meaning the fourth-highest daily 8-hour ozone level in a year. This is used instead of the highest reading so that one unusual day doesn't distort the data. The values are around 0.06–0.10 parts per million (ppm). Ozone can irritate the lungs, trigger asthma attacks, and make it harder for people to breathe. Ozone values > 0.07 ppm is considered harmful to breathing, and especially for those with asthma. Ozone values > 0.08 is harmful for everyone, and > 0.3 means reduction in lung function and chest pain. 

Another metric important to our question is PM 2.5. PM 2.5 values are measured in micrograms per cubic meter (µg/m^3). The “Weighted Annual Mean” represents the average exposure throughout the year. Some cities had averages above 15 µg/m³ in the early 2000s, but many are closer to about 7–10 in recent years. This matters because long-term exposure to PM2.5 is strongly linked to heart disease, lung cancer, and other health problems. 0-12 µg/m^3 is considered satisfactory, 12.1-13.4 µg/m^3 moderate but not good for sensitve people, and > 35 is unhealthy/hazardous. 

There are a few concerns with this dataset. Firstly, the file notes that the EPA found data quality issues affecting some states, including Florida, Illinois, most of Tennessee, and parts of Kentucky. Measurements from those areas may not be fully reliable. Also, cities have different numbers of monitoring stations, so the data could be not consistent between cities. Another limitation is that the dataset relies on summary statistics rather than raw daily readings. While this makes it easier to see long-term trends, it can hide short-term spikes such as wildfire smoke events.

```python
# 1. load dataset
air = pd.read_excel('data/00-raw/airqualitytrendsbycity.xlsx',
                    header=3)

print(air.columns.tolist())

# 2. make dataset tidy
air = air.iloc[3:].reset_index(drop=True) # drop metadata
# fill grouped labels
air["Core Based Statistical Area"] = air["Core Based Statistical Area"].ffill()
air = air.rename(columns={
    "Core Based Statistical Area":"cbsa",
    "Pollutant":"pollutant",
    "Trend Statistic":"trend_stat",
    "Number of Trends Sites":"sites"
})
#dataset has each year in it's own column, but 
#each variable should have it's own column while years should be stored in a single 'year' column
years = [c for c in air.columns if str(c).isdigit()]
air[years] = air[years].apply(pd.to_numeric, errors='coerce')

air = air.melt(
    id_vars=["cbsa","pollutant","trend_stat","sites"],
    value_vars=years,
    var_name="year",
    value_name="value"
)

#3. demonstrate size of data
print("Shape of data:", air.shape)

#4. find out how much data is missing
print("Missing data: \n", air.isna().sum())
print("\nMissing %:", air["value"].isna().mean()*100)
```

Less than 1% of the data is missing, and it seems to be from earlier years, which suggests that some cities didn't have monitoring stations yet. 

```python
#5. find outliers

def find_outliers(df):
    q1 = df.value.quantile(.25)
    q3 = df.value.quantile(.75)
    iqr = q3-q1
    return df[(df.value < q1-1.5*iqr) | (df.value > q3+1.5*iqr)]

outliers = air.groupby("pollutant")[air.columns].apply(find_outliers)

print("Outliers found:", len(outliers))
```

```python
# 6. clean the data

air = air.dropna(subset=["pollutant", "cbsa"])
print(air.isna().sum())  
air.head()
```

### San Francisco Registration and Fuel Dataset

This dataset contains counts of vehicle registrations in California by zip code and fuel type, covering the years 2018 through 2024, excluding 2019. The key metrics are:

- **Date**: the date (month/day/year) the vehicle was registered.
- **Zip Code**: the 5-digit numeric code representing the geographic area where the vehicle was registered.
- **Model** Year: the year the vehicle model was manufactured. This can sometimes be represented with ranges in case the exact same vehicle model was manufactured for multiple years.
- **Fuel**: the fuel type the vehicle utilizes.
- **Make**: the manufacturer of the vehicle.
- **Duty**: the duty type (heavy or light) of the vehicle.
- **Vehicles**: the number of vehicles corresponding to those attributes (date, zip code, model year, etc.). These values can range from 1 to several thousand per row.

**Note**: Some dataset labels do not match the actual registration dates. For example, the file labeled 2019 contained late 2018 registrations and no 2019 registrations. Moreover, the links labeled for 2023, 2024, and 2025 actually linked to the previous year's file and the 2022 dataset had both the 2023 and 2022 links directing to it. These discrepancies are noted here for transparency before analysis and directly reported to the official California Open Data portal.

```python
# files to be parsed for San Francisco registration data
sf_ev_files = [
    'data/00-raw/vehicle_fuel_2018.csv',
    'data/00-raw/vehicle_fuel_2020.csv',
    'data/00-raw/vehicle_fuel_2021.csv',
    'data/00-raw/vehicle_fuel_2022.csv',
    'data/00-raw/vehicle_fuel_2023.csv',
    'data/00-raw/vehicle_fuel_2024.csv'
]

# creating a single dataframe to combine all the datasets
sf_ev_dfs = []

for file in sf_ev_files:
    df = pd.read_csv(file, low_memory=False)
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
    # standardizing the columns so that they uniform across the different datasets
    sf_ev_dfs.append(df)

sf_ev_data = pd.concat(sf_ev_dfs, ignore_index=True)

# dropping the id column the imported csv contain for clarity
if '_id' in sf_ev_data.columns:
    sf_ev_data = sf_ev_data.drop(columns=['_id'])
```

```python
# shape and size of the dataset before any manipulation
print("Dataset shape: ", sf_ev_data.shape)
print("Entries: ", sf_ev_data.size)
sf_ev_data
```

The key columns we will focus on for analysis are date, zip_code, fuel, and vehicles. These columns contain critical information needed to examine vehicle registrations by year, location, fuel type, and count. The remaining columns are irrelevant because even if they contain missing values, they have no effect on the attributes we are analyzing, so we can safely ignore them.

Because our analysis relies on complete and accurate information for these key columns, any rows containing missing or unknown data are removed. This ensures our dataset is clean and fully reliable for analysis later on.

```python
# keep only key columns
sf_ev_clean_data = sf_ev_data[['date', 'zip_code', 'fuel', 'vehicles']].copy()

# drop rows where vehicles is missing or <= 0
sf_ev_clean_data = sf_ev_clean_data.dropna(subset=['vehicles'])
sf_ev_clean_data = sf_ev_clean_data[sf_ev_clean_data['vehicles'] > 0]

# ensure zip_code and fuel are strings
sf_ev_clean_data['zip_code'] = sf_ev_clean_data['zip_code'].astype(str)
sf_ev_clean_data['fuel'] = sf_ev_clean_data['fuel'].astype(str)

# strip spaces and lower-case
sf_ev_clean_data['zip_code'] = sf_ev_clean_data['zip_code'].str.strip().str.lower()
sf_ev_clean_data['fuel'] = sf_ev_clean_data['fuel'].str.strip().str.lower()

# remove unknown / other values
unk_values = ['unk', 'unknown', 'other', '']
sf_ev_clean_data = sf_ev_clean_data[
    ~sf_ev_clean_data['zip_code'].isin(unk_values) &
    ~sf_ev_clean_data['fuel'].isin(unk_values)
]
```

```python
sf_ev_clean_data
```

Now that we have cleaned our dataset, we can begin standardizing it to fit our analysis needs.

Extract the year from date: The date column contains full registration date, which is more detailed than necessary for our anaylsis. We will extract the year and replace the date with that.

Filtering for San Francisco zip codes: This analysis focuses specifically on San Francisco, so we will filter the dataset to include only relevant zip codes. San Francisco zip codes generally begin with 941, which allows for an easy way to filter the dataset.

Standardizing fuel types: The dataset contains many specific fuel types, some of which are rare or hybrids, so we can standardize them into broader categories such as Gasoline, Diesel, Electric, and Hybrid.

Combining the vehicle counts: Now we can combine all the vehicle counts into their respective fuel type, year, and location to get rid of duplicates or ambiguity in the rows.

```python
# extracting the year from date and renaming the column
sf_ev_clean_data['date'] = pd.to_datetime(sf_ev_clean_data['date'])
sf_ev_clean_data['year'] = sf_ev_clean_data['date'].dt.year
sf_ev_clean_data = sf_ev_clean_data.drop(columns=['date'])

# filtering for San Francisco zip codes
sf_ev_clean_data = sf_ev_clean_data[sf_ev_clean_data['zip_code'].str.startswith('941')]

# function to standardize the fuel types
def standardize_fuel(fuel):
    fuel = fuel.strip().lower()
    if fuel in ['gasoline', 'flex-fuel', 'natural gas']:
        return 'Gasoline'
    elif fuel == 'diesel and diesel hybrid':
        return 'Diesel'
    elif fuel in ['battery electric', 'hydrogen fuel cell']:
        return 'Electric'
    elif fuel in ['plug-in hybrid', 'hybrid gasoline']:
        return 'Hybrid'
    else:
        return 'Other'

sf_ev_clean_data['fuel'] = sf_ev_clean_data['fuel'].apply(standardize_fuel)

sf_ev_clean_data = (
    sf_ev_clean_data
    .groupby(['year', 'zip_code', 'fuel'], as_index=False)
    .agg({'vehicles': 'sum'})
)
```

### Aggregated Smoking & Secondhand Smoke Exposure Dataset

This dataset was compiled from three separate public health surveys to track smoking-related metrics across three U.S. cities: Harris County (TX), San Francisco County (CA), and New York City (NY). It covers years ranging from 2003 to 2024, though the time coverage varies significantly by geography.
The primary metric across most of the dataset is current smoker prevalence (%), the share of adults who reported being a current smoker at the time of the survey. For Harris County, this comes from the Harris County Community Health Survey and covers 2013–2022 (with gaps in 2018 and 2019). For San Francisco, this comes from the California Health Interview Survey (CHIS) and is the most complete series, spanning 2003–2024 with 18 data points. Values across both geographies range from about 4.7% to 14.6%, reflecting a general downward trend in smoking rates over time.

The New York City data is the exception: rather than current smoker prevalence, it measures secondhand smoke exposure at home (everyday %) — the share of adults who reported being exposed to secondhand smoke in their home every day. This comes from the NYC Environment & Health Data Portal and covers only three survey years (2013, 2014, and 2017), with values around 9–11%.

There are a few important limitations to keep in mind. The most significant is that the NYC metric is fundamentally different from the other two, secondhand smoke exposure is not the same as smoking prevalence, so NYC cannot be directly compared to Harris County or San Francisco without additional context or caveats. Year coverage is also uneven, with NYC having far fewer data points than the other two geographies, limiting any time-trend analysis for that city. Finally, 4 of the 29 estimates are flagged as statistically unstable by their original sources, meaning the sample sizes behind those estimates were small enough that the values should be interpreted with caution. This is something that will need to be considered when doing our data analysis and will have to be fixed by using another data set or accounting for these shortcomings.

```python
"""
Combines three smoking-related datasets into a single long-format CSV.

Sources:
  1. adults-who-smoke.csv           – Harris County, TX current smoker %
  2. AskCHISResults*.csv            – San Francisco County CHIS current smoker %
  3. NYC_EH_Data_Portal_*.csv       – NYC citywide secondhand smoke exposure %

Output: combined_smoking_data.csv
"""

import pandas as pd
import re

# ── 1. Harris County ──────────────────────────────────────────────────────────
harris = pd.read_csv('data/00-raw/adults-who-smoke.csv', encoding='utf-8-sig')
harris.columns = ['Year', 'Current_Smoker_Pct']
harris['Year'] = harris['Year'].astype(int)
harris['Geography'] = 'Harris County, TX'
harris['Source'] = 'Harris County CHS'
harris['Metric'] = 'Current Smoker (%)'
harris['Statistically_Unstable'] = False
harris = harris[['Year','Geography','Source','Metric','Current_Smoker_Pct','Statistically_Unstable']]
```

The date column contains full registration dates (MM/DD/YYYY). Since our analysis focuses on year-over-year trends, we extract just the year.

```python
# ── 2. San Francisco County ───────────────────────────────────────────
# parse excel into csv style
san_fran_raw = pd.read_excel('data/00-raw/AskCHISResults202602191325.csv', header=None)
san_fran_data = san_fran_raw.iloc[7:25, [0, 1]].copy()
san_fran_data.columns = ['Label', 'Pct']
san_fran_data = san_fran_data[san_fran_data['Label'].str.contains('Current Smoker', na=False)]

def parse_san_fran(label, pct):
    year = int(re.search(r'\((\d{4})\)', str(label)).group(1))
    unstable = '*' in str(pct)
    pct_val = float(re.sub(r'[^\d.]', '', str(pct)))
    return year, pct_val, unstable

rows = [parse_san_fran(r['Label'], r['Pct']) for _, r in san_fran_data.iterrows()]
san_fran = pd.DataFrame(rows, columns=['Year','Current_Smoker_Pct','Statistically_Unstable'])
san_fran['Geography'] = 'San Francisco County, CA'
san_fran['Source'] = 'CA Health Interview Survey (CHIS)'
san_fran['Metric'] = 'Current Smoker (%)'
san_fran = san_fran[['Year','Geography','Source','Metric','Current_Smoker_Pct','Statistically_Unstable']]
```

The NYC data is structured differently from the other two sources — it is stored as a proper CSV and requires filtering to citywide-level rows only (as opposed to borough or neighborhood-level). We also parse the `Everyday (percent)` column, which contains confidence intervals embedded in the string alongside the point estimate, so we extract just the numeric value and flag any statistically unstable estimates.

```python
# ── 3. NYC Secondhand Smoke (Citywide rows only) ──────────────────────────────
nyc_raw = pd.read_csv('data/00-raw/NYC EH Data Portal - Secondhand smoke at home (from outside sources) (full table).csv')
nyc_city = nyc_raw[nyc_raw['GeoType'] == 'Citywide'].copy().reset_index(drop=True)

def parse_nyc(val):
    s = str(val)
    unstable = '*' in s
    m = re.search(r'[\d.]+', s)
    pct = float(m.group()) if m else None
    return pct, unstable

nyc_pcts, nyc_unstable = [], []
for v in nyc_city['Everyday (percent)']:
    p, u = parse_nyc(v)
    nyc_pcts.append(p)
    nyc_unstable.append(u)

nyc = pd.DataFrame({
    'Year': nyc_city['TimePeriod'].astype(int),
    'Geography': 'New York City, NY',
    'Source': 'NYC EH Data Portal',
    'Metric': 'Secondhand Smoke Exposure at Home – Everyday (%)',
    'Current_Smoker_Pct': nyc_pcts,
    'Statistically_Unstable': nyc_unstable,
})
```

```python
combined = pd.concat([harris, san_fran, nyc], ignore_index=True)
combined = combined.sort_values(['Geography', 'Year']).reset_index(drop=True)
combined.to_csv('data/01-interim/combined_smoking_data.csv', index=False)
```

```python
import requests
import io

# URL (CSV format)
sf_asthma_url = "https://data.chhs.ca.gov/dataset/01f456c3-db34-44f2-a52c-6811bef8ba6d/resource/a440b99b-ccc6-473c-bea1-2baf36b05dbe/download/current-asthma-prevalence-by-county-2015_2022.csv"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(sf_asthma_url, headers=headers)

if response.status_code == 200:
    # ADD encoding='ISO-8859-1' OR encoding='cp1252'
    sf_df = pd.read_csv(io.BytesIO(response.content), encoding='ISO-8859-1')
    sf_df = sf_df.drop(['STRATA', '95% CONFIDENCE INTERVAL', 'COUNTIES GROUPED', 'COMMENT'], axis=1)

    sf_df = sf_df.rename(columns={'COUNTY':'city', 'YEARS':'year', 'AGE GROUP':'age', 'CURRENT PREVALENCE':'percentage'})
    sf_df = sf_df[sf_df['city']=='San Francisco']
    sf_df = sf_df[sf_df['age']=='18+ years']
    sf_df['city'] = 'san francisco'
    sf_df = sf_df.reset_index(drop=False)
    
    sf_df.loc[sf_df['year'].str.contains('2015'), 'year'] = '2015'
    sf_df.loc[sf_df['year'].str.contains('2017'), 'year'] = '2017'
    sf_df.loc[sf_df['year'].str.contains('2019'), 'year'] = '2019'
    sf_df.loc[sf_df['year'].str.contains('2021'), 'year'] = '2021'

    # Using the trick from before to just take the first 4 characters
    sf_df['year'] = sf_df['year'].str[:4].astype(int)
    
    # Create a copy of the dataframe to represent the "even" years
    even_years_df = sf_df.copy()
    
    # Add 1 to the year column (e.g., 2015 becomes 2016)
    even_years_df['year'] = even_years_df['year'] + 1
    
    # Combine the original "odd" years with the new "even" years
    sf_df = pd.concat([sf_df, even_years_df])
    
    # Sort by year so it's chronological and reset the index
    sf_df = sf_df.sort_values('year').reset_index(drop=True)
    sf_df = sf_df.drop(['index','age'], axis=1)
else:
    print(f"Failed to download. Status code: {response.status_code}")

sf_df
```
