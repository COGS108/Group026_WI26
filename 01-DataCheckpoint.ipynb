## Background and Prior Work

 ### Background:
Air pollution is one of the leading environmental health risks worldwide, and urban transportation is a major contributor. Private vehicles, such as cars, trucks, and motorcycles, emit pollutants such as particulate matter (PM2.5 and PM10), nitrogen dioxide (NO2), carbon monoxide (CO), and volatile organic compounds (VOCs), all of which have been strongly linked to respiratory diseases, including asthma and chronic obstructive pulmonary disease (COPD).

As cities grow, the concentration of vehicles increases, intensifying both traffic congestion and emissions, which has direct implications for public health. However, the magnitude of this effect is not fully understood. Understanding the link between transportation and air quality is crucial for policy-making aimed at improving population health.

Public transportation, such as buses, metros, and high-speed rail (HSR), has long been proposed as a solution to reduce private vehicle use. Expanding public transit can enable ease of movement for lower-income individuals while lowering urban air pollution by reducing congestion and moving more people collectively without increasing operational emissions per person.

Other factors besides transportation also influence urban air quality and lung health, including smoking rates, adoption of electric vehicles (EVs), the movement of industrial emissions from surrounding areas into urban centers, and socioeconomic factors, which can have both positive and negative effects.

Our project aims to specifically observe the link between public transportation systems and air quality and lung disease, controlling for the factors mentioned above. Some of the major cities we will investigate include Tokyo, London, New York City, Beijing, and Paris.

It is important to note that our study is observational and relies on publicly available datasets, so we cannot definitively prove causation between public transportation and air quality. Many of the factors mentioned above are interrelated, and unobservable variables may influence the results. To mitigate bias, we will examine multiple cities and control for known confounding factors as carefully as possible, aiming to produce robust, evidence-based insights.

### Air Quality Data

This dataset looks at air pollution trends in U.S. cities (called Core Based Statistical Areas, or CBSAs) from 2000 to 2023. Each entry shows a pollutant measured in a specific city along with a statistic that summarizes how that pollutant is evaluated. For instance, one major metric is ozone (O3), which is reported using the “4th Max” value — meaning the fourth-highest daily 8-hour ozone level in a year. This is used instead of the highest reading so that one unusual day doesn't distort the data. The values are around 0.06–0.10 parts per million (ppm). Ozone can irritate the lungs, trigger asthma attacks, and make it harder for people to breathe. Ozone values > 0.07 ppm is considered harmful to breathing, and especially for those with asthma. Ozone values > 0.08 is harmful for everyone, and > 0.3 means reduction in lung function and chest pain. 

Another metric important to our question is PM 2.5. PM 2.5 values are measured in micrograms per cubic meter (µg/m^3). The “Weighted Annual Mean” represents the average exposure throughout the year. Some cities had averages above 15 µg/m³ in the early 2000s, but many are closer to about 7–10 in recent years. This matters because long-term exposure to PM2.5 is strongly linked to heart disease, lung cancer, and other health problems. 0-12 µg/m^3 is considered satisfactory, 12.1-13.4 µg/m^3 moderate but not good for sensitve people, and > 35 is unhealthy/hazardous. 

There are a few concerns with this dataset. Firstly, the file notes that the EPA found data quality issues affecting some states, including Florida, Illinois, most of Tennessee, and parts of Kentucky. Measurements from those areas may not be fully reliable. Also, cities have different numbers of monitoring stations, so the data could be not consistent between cities. Another limitation is that the dataset relies on summary statistics rather than raw daily readings. While this makes it easier to see long-term trends, it can hide short-term spikes such as wildfire smoke events.

```python
# 1. load dataset
path = Path("data/00-raw/airqualitytrendsbycity.csv")
air = pd.read_csv(path, skiprows=3)

# 2. make dataset tidy
air = air.iloc[3:].reset_index(drop=True) # drop metadata
# fill grouped labels
air["Core Based Statistical Area"] = air["Core Based Statistical Area"].ffill()
air = air.rename(columns={
    "Core Based Statistical Area":"cbsa",
    "Pollutant":"pollutant",
    "Trend Statistic":"trend_stat",
    "Number of Trends Sites":"sites"
})
#dataset has each year in it's own column, but 
#each variable should have it's own column while years should be stored in a single 'year' column
years = [c for c in air.columns if str(c).isdigit()]
air[years] = air[years].apply(pd.to_numeric, errors='coerce')

air = air.melt(
    id_vars=["cbsa","pollutant","trend_stat","sites"],
    value_vars=years,
    var_name="year",
    value_name="value"
)

#3. demonstrate size of data
print("Shape of data:", air.shape)

#4. find out how much data is missing
print("Missing data: \n", air.isna().sum())
print("\nMissing %:", air["value"].isna().mean()*100)
```

Less than 1% of the data is missing, and it seems to be from earlier years, which suggests that some cities didn't have monitoring stations yet. 

```python
#5. find outliers

def find_outliers(df):
    q1 = df.value.quantile(.25)
    q3 = df.value.quantile(.75)
    iqr = q3-q1
    return df[(df.value < q1-1.5*iqr) | (df.value > q3+1.5*iqr)]

outliers = air.groupby("pollutant")[air.columns].apply(find_outliers)

print("Outliers found:", len(outliers))
```

```python
# 6. clean the data

air = air.dropna(subset=["pollutant", "cbsa"])
print(air.isna().sum())  
air.head()
```

### San Francisco Registration and Fuel Dataset

This dataset contains counts of vehicle registrations in California by zip code and fuel type, covering the years 2018 through 2024, excluding 2019. The key metrics are:

- **Date**: the date (month/day/year) the vehicle was registered.
- **Zip Code**: the 5-digit numeric code representing the geographic area where the vehicle was registered.
- **Model** Year: the year the vehicle model was manufactured. This can sometimes be represented with ranges in case the exact same vehicle model was manufactured for multiple years.
- **Fuel**: the fuel type the vehicle utilizes.
- **Make**: the manufacturer of the vehicle.
- **Duty**: the duty type (heavy or light) of the vehicle.
- **Vehicles**: the number of vehicles corresponding to those attributes (date, zip code, model year, etc.). These values can range from 1 to several thousand per row.

**Note**: Some dataset labels do not match the actual registration dates. For example, the file labeled 2019 contained late 2018 registrations and no 2019 registrations. Moreover, the links labeled for 2023, 2024, and 2025 actually linked to the previous year's file and the 2022 dataset had both the 2023 and 2022 links directing to it. These discrepancies are noted here for transparency before analysis and directly reported to the official California Open Data portal.

```python
# files to be parsed for San Francisco registration data
sf_ev_files = [
    'data/00-raw/vehicle_fuel_2018.csv',
    'data/00-raw/vehicle_fuel_2020.csv',
    'data/00-raw/vehicle_fuel_2021.csv',
    'data/00-raw/vehicle_fuel_2022.csv',
    'data/00-raw/vehicle_fuel_2023.csv',
    'data/00-raw/vehicle_fuel_2024.csv'
]

# creating a single dataframe to combine all the datasets
sf_ev_dfs = []

for file in sf_ev_files:
    df = pd.read_csv(file, low_memory=False)
    df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
    # standardizing the columns so that they uniform across the different datasets
    sf_ev_dfs.append(df)

sf_ev_data = pd.concat(sf_ev_dfs, ignore_index=True)

# dropping the id column the imported csv contain for clarity
if '_id' in sf_ev_data.columns:
    sf_ev_data = sf_ev_data.drop(columns=['_id'])
```

```python
# shape and size of the dataset before any manipulation
print("Dataset shape: ", sf_ev_data.shape)
print("Entries: ", sf_ev_data.size)
sf_ev_data
```

The key columns we will focus on for analysis are date, zip_code, fuel, and vehicles. These columns contain critical information needed to examine vehicle registrations by year, location, fuel type, and count. The remaining columns are irrelevant because even if they contain missing values, they have no effect on the attributes we are analyzing, so we can safely ignore them.

Because our analysis relies on complete and accurate information for these key columns, any rows containing missing or unknown data are removed. This ensures our dataset is clean and fully reliable for analysis later on.

```python
# keep only key columns
sf_ev_clean_data = sf_ev_data[['date', 'zip_code', 'fuel', 'vehicles']].copy()

# drop rows where vehicles is missing or <= 0
sf_ev_clean_data = sf_ev_clean_data.dropna(subset=['vehicles'])
sf_ev_clean_data = sf_ev_clean_data[sf_ev_clean_data['vehicles'] > 0]

# ensure zip_code and fuel are strings
sf_ev_clean_data['zip_code'] = sf_ev_clean_data['zip_code'].astype(str)
sf_ev_clean_data['fuel'] = sf_ev_clean_data['fuel'].astype(str)

# strip spaces and lower-case
sf_ev_clean_data['zip_code'] = sf_ev_clean_data['zip_code'].str.strip().str.lower()
sf_ev_clean_data['fuel'] = sf_ev_clean_data['fuel'].str.strip().str.lower()

# remove unknown / other values
unk_values = ['unk', 'unknown', 'other', '']
sf_ev_clean_data = sf_ev_clean_data[
    ~sf_ev_clean_data['zip_code'].isin(unk_values) &
    ~sf_ev_clean_data['fuel'].isin(unk_values)
]
```

```python
sf_ev_clean_data
```

Now that we have cleaned our dataset, we can begin standardizing it to fit our analysis needs.

Extract the year from date: The date column contains full registration date, which is more detailed than necessary for our anaylsis. We will extract the year and replace the date with that.

Filtering for San Francisco zip codes: This analysis focuses specifically on San Francisco, so we will filter the dataset to include only relevant zip codes. San Francisco zip codes generally begin with 941, which allows for an easy way to filter the dataset.

Standardizing fuel types: The dataset contains many specific fuel types, some of which are rare or hybrids, so we can standardize them into broader categories such as Gasoline, Diesel, Electric, and Hybrid.

Combining the vehicle counts: Now we can combine all the vehicle counts into their respective fuel type, year, and location to get rid of duplicates or ambiguity in the rows.

```python
# extracting the year from date and renaming the column
sf_ev_clean_data['date'] = pd.to_datetime(sf_ev_clean_data['date'])
sf_ev_clean_data['year'] = sf_ev_clean_data['date'].dt.year
sf_ev_clean_data = sf_ev_clean_data.drop(columns=['date'])

# filtering for San Francisco zip codes
sf_ev_clean_data = sf_ev_clean_data[sf_ev_clean_data['zip_code'].str.startswith('941')]

# function to standardize the fuel types
def standardize_fuel(fuel):
    fuel = fuel.strip().lower()
    if fuel in ['gasoline', 'flex-fuel', 'natural gas']:
        return 'Gasoline'
    elif fuel == 'diesel and diesel hybrid':
        return 'Diesel'
    elif fuel in ['battery electric', 'hydrogen fuel cell']:
        return 'Electric'
    elif fuel in ['plug-in hybrid', 'hybrid gasoline']:
        return 'Hybrid'
    else:
        return 'Other'

sf_ev_clean_data['fuel'] = sf_ev_clean_data['fuel'].apply(standardize_fuel)

sf_ev_clean_data = (
    sf_ev_clean_data
    .groupby(['year', 'zip_code', 'fuel'], as_index=False)
    .agg({'vehicles': 'sum'})
)
```
