## Research Question

How does public transportation usage (buses and metro), smoking prevalence rates, and electric vehicle (EV) adoption influence air quality and the rate of reported asthma attacks among asthma patients in Houston, San Francisco (SF), and New York City (NYC)?

## Background and Prior Work

### Background:
Air pollution is one of the leading environmental health risks worldwide, and urban transportation is a major contributor. Private vehicles, such as cars, trucks, and motorcycles, emit pollutants such as particulate matter (PM2.5 and PM10), nitrogen dioxide (NO2), carbon monoxide (CO), and volatile organic compounds (VOCs), all of which have been strongly linked to respiratory diseases, including asthma and chronic obstructive pulmonary disease (COPD). (1, 2)

From the report of Asthma and Allergy Foundation of America (5), poor air quality can negatively affect everyone’s health. Research shows that air pollution can make asthma worse and trigger asthma symptoms. Air pollution includes gases, smoke from fires, volcanic ash, dust particles, emissions from transportation, and other substances that can harm the lungs.

As cities grow, the concentration of vehicles increases, intensifying both traffic congestion and emissions, which has direct implications for public health. Understanding the link between transportation and air quality is crucial for policy-making aimed at improving population health.

Public transportation, such as buses, metros, and high-speed rail (HSR), has long been proposed as a solution to reduce private vehicle use. Expanding public transit can enable ease of movement for lower-income individuals while lowering urban air pollution by reducing congestion and moving more people collectively without increasing operational emissions per person. (3)

Other factors besides transportation also influence urban air quality and lung health, including smoking rates and adoption of EVs, which can have both positive and negative effects.

Our project aims to specifically observe the link between public transportation systems and air quality and lung disease, controlling for the factors mentioned above. Some of the cities we will investigate include Houston, NYC, and SF.

It is important to note that our study is observational and relies on publicly available datasets, so we cannot definitively prove causation between public transportation and air quality. Many of the factors mentioned above are interrelated, and unobservable variables may influence the results. To mitigate bias, we will examine multiple cities and control for known confounding factors as carefully as possible, aiming to produce robust, evidence-based insights.


### Prior Work:
A relevant prior work is the technical report *Assessing the Impact of Air Pollution on Public Health along Transit Routes*, which examines how transportation and socioeconomic factors influence respiratory health in some major metropolitan areas, Dallas-Fort Worth and LA. The researchers applied stat techniques such as PCA and OLS to identify predictors of respiratory disease. Their findings showed strong correlations between both demographics and access to transit/automobile access to respiratory disease risk. Interestingly, populations with higher transit access were more prone to respiratory diseases. This work relates closely to our proposed project because it demonstrates a clear connection between transportation systems, air quality, and respiratory health. (3)

Another study published in Environmental Health Perspectives investigated the association between traffic-related air pollution and the incidence of asthma in children living near major roadways. The researchers found that exposure to combustion-related pollutants such as nitrogen dioxide (NO2) and ozone was significantly linked to increases in asthma prevalence among children at home and school environments. By analyzing air quality data alongside health outcomes, the study showed that proximity to heavy traffic corridors is a significant predictor of respiratory health risks, particularly for vulnerable populations such as children. These findings highlight how motor vehicle emissions can contribute to asthma development, supporting the our rationale for investigating transportation factors alongside other predictors like smoking prevalence and EV usage in our project. (4)

- #### Dataset #1

  - **Link to the dataset:** https://lab.data.ca.gov/dataset/vehicle-fuel-type-count-by-zip-code
  - **Number of observations:** ~4,000,000 rows (raw), 42 rows after cleaning and filtering for SF (6 fuel types × 7 years)
  - **Number of variables:** 7 raw (date, zip_code, fuel, vehicles, make, model_year, duty), 4 after cleaning (fuel_type, year, vehicles, percentage)

  - **Descriptions of any shortcomings this dataset has with respect to the project:** The California DMV does not release actual registration dates and instead standardizes them to a single date per year. The 2018 dataset only records through 1/10/2018, meaning some 2018 registrations appear in the 2019 file. Additionally, the raw dataset contains many granular and inconsistent fuel type labels that required significant normalization before analysis.

- #### Dataset #2
  - **Dataset Name:** Texas Alternatively Fueled Vehicle Registration Dataset
  - **Link to the dataset:** https://www.txdmv.gov/sites/default/files/report-files/FY-2023-Alternatively-Fueled-Vehicle-Report.pdf
  - **Number of observations:** 144 (18 fuel types × 8 years)
  - **Number of variables:** 3 (fuel_type, year, registered_vehicles)
  - **Description of the variables most relevant to this project:** fuel_type (the type of fuel used by the vehicle), year (fiscal year 2016–2023), and registered_vehicles (the total number of registered vehicles matching that fuel type for that year).
  - **Descriptions of any shortcomings this dataset has with respect to the project:** The TxDMV does not publish county-level or city-level breakdowns by fuel type, meaning Houston-specific data is not publicly available. Texas state-level data is used as a proxy for Houston, which is a limitation given that statewide trends may not perfectly reflect Houston's urban vehicle composition. Additionally, the dataset only covers alternative fuels in detail, with gasoline and other conventional fuels reported as aggregate totals.

- #### Dataset #3
  - **Dataset Name:** U.S. Vehicle Registration Counts by State and Fuel Type (AFDC)
  - **Link to the dataset:** https://afdc.energy.gov/vehicle-registration
  - **Number of observations:** 468 (52 states/territories × 9 years)
  - **Number of variables:** 14 raw (state, year, and one column per fuel type), 4 after cleaning (fuel_type, year, vehicles, percentage)
  - **Description of the variables most relevant to this project:** state (filtered to New York as a proxy for NYC), year (2016–2024), and per-fuel-type counts including electric, plug-in hybrid, hybrid electric, gasoline, diesel, flex, and others.
  - **Descriptions of any shortcomings this dataset has with respect to the project:** The AFDC dataset is only available at the state level, meaning NYC-specific registration data by fuel type is not available from any public source. New York state is used as a proxy for NYC, which is a limitation since the state includes large rural areas with different vehicle composition than NYC. The NY Open Data Portal DMV dataset was initially considered as a more granular alternative but was found to be unsuitable for year-over-year analysis, as it is a current snapshot of active registrations rather than a historical record, causing severe undercounting in earlier years and artificial inflation in recent years.

- #### Dataset #4
  - **Dataset Name:** FTA National Transit Database (NTD) Monthly Ridership — Calendar Year UPT
  - **Link to the dataset:** https://www.transit.dot.gov/ntd/data-product/monthly-module-adjusted-data-release
  - **Number of observations:** 72 (3 cities × 24 years, 2002–2025)
  - **Number of variables:** 3 after cleaning (year, ridership, city)
  - **Description of the variables most relevant to this project:** year (calendar year 2002–2025), ridership (total unlinked passenger trips aggregated across all transit modes and service types for each city's urbanized area), and city (Houston, NYC, or SF).
  - **Descriptions of any shortcomings this dataset has with respect to the project:** The NTD reports ridership at the urbanized area (UZA) level rather than the city boundary level, meaning each UZA includes surrounding suburbs beyond the city limits. Additionally, ridership figures are aggregated across all modes and service types (bus, demand response, commuter bus, etc.), so mode-level breakdowns are lost after aggregation. NaN values in the raw data, indicating a mode was not yet operational or not reported in a given year, are treated as zero in the sum, which may slightly undercount ridership in early years when reporting coverage was less complete.

For instance, NO2 is measured by “Annual Mean”, while PM 2.5 is measured by “Weighted Annual Mean”. The difference in measurement might mean standardizing or further processing of these statistics, or looking at these two pollutants separately. ### EDA: Houston Multi-Variable Correlation and Trends (2015–2022)
This analysis aims to explore how smoking rate, public transport ridership, and EV adoption affect asthma rate and air quality in Houston between 2015–2022.

The goal of this EDA is to:

1. **Identify Co-occurrences:** Examine the pairwise correlation between each variable.
2. **Environmental Transition:** Compare growth in EV adoption and public transport ridership as indicators of environmental awareness.
3. **Correlation Between the Two Indicators:** Examine whether improvement in air quality (lower PM2.5) and drop in asthma rate have a negative correlation.
