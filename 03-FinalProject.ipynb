## Abstract

This project investigated whether public transportation usage, EV adoption, and smoking prevalence are associated with air quality and asthma rates in Houston, San Francisco, and New York City. We hypothesized that higher transit ridership and EV adoption, combined with lower smoking rates, would predict better air quality and fewer asthma episodes and that transit would be the strongest predictor of the three.

To test this, we collected annual data from 2015–2022 across all three cities and ran OLS regression models both pooled across cities and separately for each one, alongside correlation analyses for each city.

Our results did not support the hypothesis. While the pooled model had a high R-squared (0.934), this was almost entirely driven by city fixed effects rather than any of our predictor variables, none of which reached statistical significance. NYC showed the most promising signal, with a strong correlation between PM2.5 and asthma rates, but SF and Houston were muddied by data quality issues and confounding time trends. Ultimately, eight years of data per city was not enough to draw reliable conclusions, and we think the next step would be moving to monthly data or looking at possible confounding factors to get the analysis needed to actually test these relationships properly.

## Background and Prior Work

### Background:
Air pollution is one of the leading environmental health risks worldwide, and urban transportation is a major contributor. Private vehicles, such as cars, trucks, and motorcycles, emit pollutants such as particulate matter (PM2.5 and PM10), nitrogen dioxide (NO2), carbon monoxide (CO), and volatile organic compounds (VOCs), all of which have been strongly linked to respiratory diseases, including asthma and chronic obstructive pulmonary disease (COPD). (1, 2)

From the report of Asthma and Allergy Foundation of America (5), poor air quality can negatively affect everyone’s health. Research shows that air pollution can make asthma worse and trigger asthma symptoms. Air pollution includes gases, smoke from fires, volcanic ash, dust particles, emissions from transportation, and other substances that can harm the lungs.

As cities grow, the concentration of vehicles increases, intensifying both traffic congestion and emissions, which has direct implications for public health. Understanding the link between transportation and air quality is crucial for policy-making aimed at improving population health.

Public transportation, such as buses, metros, and high-speed rail (HSR), has long been proposed as a solution to reduce private vehicle use. Expanding public transit can enable ease of movement for lower-income individuals while lowering urban air pollution by reducing congestion and moving more people collectively without increasing operational emissions per person. (3)

Other factors besides transportation also influence urban air quality and lung health, including smoking rates and adoption of EVs, which can have both positive and negative effects.

Our project aims to specifically observe the link between public transportation systems and air quality and lung disease, controlling for the factors mentioned above. Some of the cities we will investigate include Houston, NYC, and SF.

It is important to note that our study is observational and relies on publicly available datasets, so we cannot definitively prove causation between public transportation and air quality. Many of the factors mentioned above are interrelated, and unobservable variables may influence the results. To mitigate bias, we will examine multiple cities and control for known confounding factors as carefully as possible, aiming to produce robust, evidence-based insights.

Their findings showed strong correlations between both demographics and access to transit/automobile access to respiratory disease risk. Interestingly, populations with higher transit access were more prone to respiratory diseases. By analyzing air quality data alongside health outcomes, the study showed that proximity to heavy traffic corridors is a significant predictor of respiratory health risks, particularly for vulnerable populations such as children. These findings highlight how motor vehicle emissions can contribute to asthma development, supporting the our rationale for investigating transportation factors alongside other predictors like smoking prevalence and EV usage in our project. The NY Open Data Portal DMV dataset was initially considered as a more granular alternative but was found to be unsuitable for year-over-year analysis, as it is a current snapshot of active registrations rather than a historical record, causing severe undercounting in earlier years and artificial inflation in recent years.

- #### Dataset #4
  - **Dataset Name:** FTA National Transit Database (NTD) Monthly Ridership — Calendar Year UPT
  - **Link to the dataset:** https://www.transit.dot.gov/ntd/data-product/monthly-module-adjusted-data-release
  - **Number of observations:** 72 (3 cities × 24 years, 2002–2025)
  - **Number of variables:** 3 after cleaning (year, ridership, city)
  - **Description of the variables most relevant to this project:** year (calendar year 2002–2025), ridership (total unlinked passenger trips aggregated across all transit modes and service types for each city's urbanized area), and city (Houston, NYC, or SF).
  - **Descriptions of any shortcomings this dataset has with respect to the project:** The NTD reports ridership at the urbanized area (UZA) level rather than the city boundary level, meaning each UZA includes surrounding suburbs beyond the city limits. Additionally, ridership figures are aggregated across all modes and service types (bus, demand response, commuter bus, etc.), so mode-level breakdowns are lost after aggregation. NaN values in the raw data, indicating a mode was not yet operational or not reported in a given year, are treated as zero in the sum, which may slightly undercount ridership in early years when reporting coverage was less complete.

### EDA: Houston Multi-Variable Correlation and Trends (2015–2022)
This analysis aims to explore how smoking rate, public transport ridership, and EV adoption affect asthma rate and air quality in Houston between 2015–2022.

The goal of this EDA is to:

1. **Identify Co-occurrences:** Examine the pairwise correlation between each variable.
2. **Environmental Transition:** Compare growth in EV adoption and public transport ridership as indicators of environmental awareness.
3. **Correlation Between the Two Indicators:** Examine whether improvement in air quality (lower PM2.5) and drop in asthma rate have a negative correlation.
4. **Smoking and Air Quality:** Examine whether smoking rate and PM2.5 move together over time.

#### Evaluation of Houston Analysis
Based on the multi-variable dataset for Houston, we can draw the following conclusions:

1. **PM2.5 Data Quality Issue**

## Conclusion

This project asked whether public transit usage, EV adoption, and smoking prevalence are associated with air quality and asthma rates across Houston, San Francisco, and New York City  and whether transit would prove the stronger predictor compared to EVs and smoking. Our hypothesis about transit being the dominant predictor could not be evaluated meaningfully under these conditions.

The city-level results gave us a few interpretable signals mixed with a lot of noise. SF presented a paradox: smoking dropped dramatically and pandemic-era air quality improved, yet asthma rates stayed elevated, likely because wildfire smoke isn't well captured in annual PM2.5 averages. Houston's PM2.5 data appeared to flatline for six consecutive years, a likely data artifact that undermines any analysis involving air quality for that city. 

Transit's role was statistically insignificant in every model, though this almost certainly reflects data limitations rather than a true null effect.

The most consistent finding across all three cities was that EV adoption and smoking rates are inversely correlated not because EVs cause people to quit smoking, but because both trends are moving in the same direction over time. This is a time-trend confound and a reminder that with only eight annual observations per city, distinguishing genuine associations from shared historical drift is essentially impossible.

The core limitation of this work is statistical power. Eight observations per city leaves only four residual degrees of freedom per model, making standard errors enormous and significance nearly unattainable even if real effects exist. Longer time series, monthly rather than annual data, or neighborhood-level observations within a single city would all substantially improve the ability to test these hypotheses. 

Future work should also consider explicitly modeling wildfire events as a covariate for SF, and sourcing more reliable PM2.5 records for Houston. A difference in modeling would move the analysis from association toward something closer to causal inference. But, again, the most crucial thing our question lacks is the data needed to analyze.
